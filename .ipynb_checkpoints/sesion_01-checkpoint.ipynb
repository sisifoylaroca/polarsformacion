```python
"""Sesión 01 de formación en Polars: importaciones básicas.

Librerías utilizadas en el notebook:

- ``datetime``: construcción de fechas y horas para generar rangos temporales.
- ``numpy`` (``np``): generación de datos aleatorios (p. ej. distribución normal).
- ``plotly.express`` (``px``): visualización interactiva de las series.
- ``polars`` (``pl``): creación y manipulación de DataFrames.
"""

import datetime

import numpy as np
import plotly.express as px
import polars as pl
import polars.selectors as cs
```

```python
from functime.cross_validation import train_test_split
from functime.seasonality import add_fourier_terms
from functime.forecasting import linear_model
from functime.preprocessing import scale
from functime.metrics import mase
```

```python
# Load commodities price data
y = pl.read_parquet("./data/commodities.parquet")
entity_col, time_col = y.columns[:2]
```

```python
# Time series split
y_train, y_test = y.pipe(train_test_split(test_size=3))
```

```python
# Fit-predict
forecaster = linear_model(freq="1mo", lags=24)
forecaster.fit(y=y_train)
y_pred = forecaster.predict(fh=3)
```

```python
# Gráfico: histórico reciente + valores reales de test + predicción
commodities = ["Aluminum", "Banana, Europe", "Tea, Mombasa", "Coffee, Arabica"]
meses_historia = 36


def preparar(df: pl.DataFrame | pl.LazyFrame, etiqueta: str) -> pl.DataFrame:
    """Filtra las commodities elegidas, unifica el tipo de fecha y etiqueta la serie."""
    return (
        df.lazy()
        .filter(pl.col(entity_col).is_in(commodities))
        .with_columns(
            pl.col(time_col).cast(pl.Datetime("us")),
            serie=pl.lit(etiqueta),
        )
        .collect()
    )


historico = y_train.lazy().sort(time_col).group_by(entity_col).tail(meses_historia)

df_plot = pl.concat(
    [
        preparar(historico, "Histórico (train)"),
        preparar(y_test, "Real (test)"),
        preparar(y_pred, "Predicción"),
    ]
)

fig = px.line(
    df_plot,
    x=time_col,
    y="price",
    color="serie",
    facet_col=entity_col,
    facet_col_wrap=2,
    markers=True,
    title="Predicción de precios de commodities (linear_model, lags=24, fh=3)",
    color_discrete_map={
        "Histórico (train)": "#636EFA",
        "Real (test)": "#00CC96",
        "Predicción": "#EF553B",
    },
    height=700,
)
fig.update_yaxes(
    matches=None, showticklabels=True
)  # cada commodity con su propia escala
fig.for_each_annotation(
    lambda a: a.update(text=a.text.split("=")[-1])
)  # títulos limpios
fig
```
